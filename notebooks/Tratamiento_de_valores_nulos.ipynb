```python
import pandas as pd
import numpy as np

# conjunto de datos
data = {
    "Producto": ["Laptop", "Teclado", "Mouse", "Monitor", "Impresora", np.nan, "Webcam", "Auriculares", "Altavoces", "Router",
                 "Teclado", "Mouse", np.nan, "Monitor", "Laptop", "Impresora", "Webcam", np.nan, "Auriculares", "Altavoces"],
    "Precio": [1200, 75, 25, 300, np.nan, 50, 60, 150, np.nan, 80,
               70, 20, 350, 280, np.nan, 45, 55, 160, 90, np.nan],
    "Stock": [50, 150, 200, 30, 10, 80, np.nan, 120, 40, 70,
              130, 180, 25, np.nan, 15, 90, 65, 110, np.nan, 55],
    "Categoria": ["Electrónica", "Periféricos", "Periféricos", "Electrónica", "Electrónica", "Periféricos",
                  "Periféricos", "Electrónica", "Electrónica", "Redes",
                  "Periféricos", "Periféricos", "Electrónica", "Electrónica", "Electrónica", "Periféricos",
                  "Periféricos", "Electrónica", "Electrónica", "Redes"],
    "Valoracion": [4.5, 4.0, 4.2, 4.8, 3.5, 4.1, 3.9, np.nan, 4.6, 4.3,
                   3.8, 4.0, np.nan, 4.7, 4.4, 3.6, 3.7, 4.5, 4.1, np.nan]
}

df = pd.DataFrame(data)
df.head(10)
```

Hacer un conteo inicial de nulos.

```python
# Contar los valores nulos por columna
print("Conteo de valores nulos por columna en df:")
df.isnull().sum().sort_values(ascending=False)
```

```python
#Porcentaje de nulos en relacion al total
print("\nPorcentaje de valores nulos en relacion al total:")
(df.isnull().mean() * 100).round(2)
```

Aplicar al menos 2 métodos de imputación distintos

```python
# Método 1: Rellenar valores nulos en 'Producto' con 'Desconocido'
df.fillna({'Producto' : "Desconocido"}, inplace = True)
df.head()
```

```python
# Método 2: Rellenar valores nulos en 'Precio' con la mediana

mediana_precio = df['Precio'].median()
df.fillna({'Precio': mediana_precio},inplace = True)

print("DataFrame después de rellenar nulos en 'Precio':")
df.head()
```

```python
# Método 3: Rellenar valores nulos en 'Stock' con la media (promedio)
media_stock = df['Stock'].mean()
df.fillna({'Stock': media_stock},inplace = True)

print("DataFrame después de rellenar nulos en 'Stock':")
df.head()
```

```python
# Método 4: Rellenar valores nulos en 'Valoracion' con la moda
moda_valoracion = df['Valoracion'].mode()[0]
df.fillna({'Valoracion': moda_valoracion},inplace = True)

print("DataFrame después de rellenar nulos en 'Valoracion':")
df.head()
```
